### Bài 9. Gradient Descent tìm hệ số đa thức

Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

# Tạo dữ liệu mẫu
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

# Khởi tạo trọng số và dữ liệu PyTorch
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

# Vòng lặp huấn luyện
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
```

### Bài 10. Giải hệ phương trình bằng Autograd

Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())
```

```python
print("\nKIỂM TRA NGHIỆM")
print(f"A + B = {A.item() + B.item():.4f} (Kỳ vọng: 9)")
print(f"C - D = {C.item() - D.item():.4f} (Kỳ vọng: 1)")
print(f"A + C = {A.item() + C.item():.4f} (Kỳ vọng: 8)")
print(f"B - D = {B.item() - D.item():.4f} (Kỳ vọng: 2)")
```

• Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

```python
import torch
torch.manual_seed(0)
print("PHẦN 1: Tensor X và các phép reshape")
 
# Tensor X có 24 phần tử (0..23)
X = torch.arange(24)
print("X ban đầu (1D, 24 phần tử):")
print(X)
 
X_4x6 = X.reshape(4, 6)
print("\nReshape thành 4x6:")
print(X_4x6)
 
X_2x12 = X.reshape(2, 12)
print("\nReshape thành 2x12:")
print(X_2x12)
 
X_2x3x4 = X.reshape(2, 3, 4)
print("\nReshape thành 2x3x4:")
print(X_2x3x4)
```

```python
print("PHẦN 2: Tensor A, B kích thước 3x4")
 
A = torch.randint(1, 10, (3, 4)).float()
B = torch.randint(1, 10, (3, 4)).float()
 
print("A =")
print(A)
print("\nB =")
print(B)
print("\nA + B =")
print(A + B)
print("\nA * B (element-wise) =")
print(A * B)
print("\nA / B (element-wise) =")
print(A / B)
# Trung bình và độ lệch chuẩn theo cột (dim=0 nghĩa là tính dọc theo từng cột)
mean_col_A = A.mean(dim=0)
std_col_A = A.std(dim=0)
mean_col_B = B.mean(dim=0)
std_col_B = B.std(dim=0)
print("\nTrung bình theo cột của A:", mean_col_A)
print("Độ lệch chuẩn theo cột của A:", std_col_A)
print("Trung bình theo cột của B:", mean_col_B)
print("Độ lệch chuẩn theo cột của B:", std_col_B)
```

```python
print("PHẦN 3: Autograd - đạo hàm y = 2x^2 + 5x + 3 tại x = 2")
 
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
 
dy_dx_autograd = x.grad.item()
# Đạo hàm thủ công: dy/dx = 4x + 5  => tại x = 2: 4*2 + 5 = 13
dy_dx_manual = 4 * 2 + 5
 
print(f"y = 2x^2 + 5x + 3 tại x = 2 -> y = {y.item()}")
print(f"dy/dx (autograd) tại x = 2  = {dy_dx_autograd}")
print(f"dy/dx (tính tay: 4x + 5)    = {dy_dx_manual}")
print("Khớp nhau!" if abs(dy_dx_autograd - dy_dx_manual) < 1e-6 else "Không khớp!")
```

```python
print("PHẦN 4: Thay đa thức Bài 9 bằng y = 2x^2 - 3x + 5")
 
import numpy as np
 
# Tạo dữ liệu mẫu theo đa thức y = 2x^2 - 3x + 5
polynomial = np.poly1d([2, -3, 5])
N = 20
X_np = np.random.randn(N, 1) * 5
Y_np = polynomial(X_np)
 
XX = np.hstack([
    X_np * X_np,
    X_np,
    np.ones_like(X_np)
])
 
# Khởi tạo trọng số và dữ liệu PyTorch
w4 = torch.randn(3, 1, requires_grad=True)
x4 = torch.tensor(XX, dtype=torch.float32)
y4 = torch.tensor(Y_np, dtype=torch.float32)
optimizer4 = torch.optim.NAdam([w4], lr=0.01)
 
print("Hệ số khởi tạo (initial coefficients):")
print(w4)
 
# Vòng lặp huấn luyện
for epoch in range(1000):
    optimizer4.zero_grad()
    y_pred4 = x4 @ w4
    mse4 = torch.mean(torch.square(y4 - y_pred4))
    mse4.backward()
    optimizer4.step()
 
print("\nHệ số học được (learned coefficients):")
print(w4)
 
a_hat, b_hat, c_hat = w4.detach().squeeze().tolist()
print("\nKết quả:")
print(f"Đa thức thật     : y = 2x^2 - 3x + 5")
print(f"Đa thức học được : y = {a_hat:.3f}x^2 + {b_hat:.3f}x + {c_hat:.3f}")
```
