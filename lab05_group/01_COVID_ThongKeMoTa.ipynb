```python
import numpy as np
import pandas as pd
from scipy import stats

covid_data = pd.read_csv("data/covid-data.csv")

covid_data.head()
```

```python
# Chọn các cột theo yêu cầu 
covid_data = covid_data[
    ["iso_code", "continent", "location", "date", "total_cases", "new_cases"]
].copy()

print("Kích thước dữ liệu:", covid_data.shape)

print("\nKiểu dữ liệu:")
print(covid_data.dtypes)

print("\nSố giá trị thiếu:")
print(covid_data.isna().sum())

covid_data.head()
```

```python
# Lấy các giá trị new_cases không bị thiếu
new_cases = covid_data["new_cases"].dropna()

print("Số quan sát ban đầu:", len(covid_data))
print("Số giá trị thiếu đã bỏ qua:", covid_data["new_cases"].isna().sum())
print("Số quan sát dùng để tính thống kê:", len(new_cases))
print("Số giá trị âm:", (new_cases < 0).sum())
```

```python
# Trung bình
data_mean = np.mean(new_cases)

# Trung vị
data_median = np.median(new_cases)

print(f"Trung bình (mean): {data_mean:,.2f}")
print(f"Trung vị (median): {data_median:,.2f}")
```

```python
# Tính mode và số lần xuất hiện
data_mode = stats.mode(new_cases.to_numpy(), keepdims=False)

print(f"Mode: {data_mode.mode:,.0f}")
print(f"Số lần xuất hiện: {data_mode.count:,}")
print(f"Tỷ lệ xuất hiện: {data_mode.count / len(new_cases):.2%}")
```

```python
# Dùng ddof=0 theo cách tính của NumPy trong bài mẫu
data_variance = np.var(new_cases, ddof=0)
data_sd = np.std(new_cases, ddof=0)

print(f"Phương sai (variance): {data_variance:,.2f}")
print(f"Độ lệch chuẩn (standard deviation): {data_sd:,.2f}")
```

```python
data_min = np.min(new_cases)
data_max = np.max(new_cases)
data_range = data_max - data_min

print(f"Giá trị nhỏ nhất: {data_min:,.0f}")
print(f"Giá trị lớn nhất: {data_max:,.0f}")
print(f"Phạm vi: {data_range:,.0f}")
```

```python
# Xem 5 dòng có new_cases lớn nhất
covid_data.nlargest(5, "new_cases")[
    ["iso_code", "continent", "location", "date", "new_cases"]
]
```

```python
# Phân vị 60% theo yêu cầu bài mẫu
data_percentile = np.percentile(new_cases, 60)

# Ba tứ phân vị
q1, q2, q3 = np.quantile(new_cases, [0.25, 0.50, 0.75])

# Khoảng tứ phân vị
data_IQR = stats.iqr(new_cases)

print(f"Phân vị 60%: {data_percentile:,.2f}")
print(f"Q1 (25%): {q1:,.2f}")
print(f"Q2 (50%): {q2:,.2f}")
print(f"Q3 (75%): {q3:,.2f}")
print(f"IQR = Q3 - Q1: {data_IQR:,.2f}")
```
