```python
import pandas as pd

marketing_data = pd.read_csv(
    "data/marketing_campaign.csv",
    sep="\t"
)

print("Kích thước dữ liệu:", marketing_data.shape)

marketing_data.head()
```

```python
marketing_data = marketing_data[
    [
        "ID", "Year_Birth", "Education", "Marital_Status",
        "Income", "Kidhome", "Teenhome", "Dt_Customer",
        "Recency", "NumStorePurchases", "NumWebVisitsMonth"
    ]
].copy()

print("Kích thước sau khi chọn cột:", marketing_data.shape)

print("\nKiểu dữ liệu:")
print(marketing_data.dtypes)

print("\nSố giá trị thiếu:")
print(marketing_data.isna().sum())

print("\nSố dòng trùng hoàn toàn:", marketing_data.duplicated().sum())

marketing_data.head()
```

```python
# Loại bỏ các dòng trùng hoàn toàn
marketing_data_duplicate = marketing_data.drop_duplicates()

print("Kích thước sau khi bỏ trùng:", marketing_data_duplicate.shape)

# Minh họa xóa dòng có nhãn index là 1
marketing_drop_row = marketing_data.drop(labels=[1], axis=0)

print("Kích thước sau khi xóa dòng index 1:", marketing_drop_row.shape)

# Minh họa xóa cột Year_Birth
marketing_drop_column = marketing_data.drop(
    labels=["Year_Birth"],
    axis=1
)

print("Kích thước sau khi xóa cột Year_Birth:", marketing_drop_column.shape)

print("Kích thước marketing_data hiện tại:", marketing_data.shape)
```

```python
# Kiểm tra các giá trị trước khi thay thế
print("Teenhome ban đầu:")
print(marketing_data["Teenhome"].value_counts().sort_index())

# Tạo cột mới, giữ lại cột gốc
marketing_data["Teenhome_replaced"] = marketing_data["Teenhome"].replace(
    [0, 1, 2],
    ["has no teen", "has teen", "has teen"]
)

print("\nSau khi thay thế:")
print(marketing_data["Teenhome_replaced"].value_counts())

marketing_data[["ID", "Teenhome", "Teenhome_replaced"]].head()
```

```python
# Giữ bản trước khi điền để dùng ở bước dropna()
marketing_before_fill = marketing_data.copy()

# Ghi nhớ những dòng Income bị thiếu
income_missing = marketing_data["Income"].isna()

# Điền giá trị thiếu bằng 0 theo bài mẫu
marketing_data["Income"] = marketing_data["Income"].fillna(0)

# Tạo cột thu nhập kiểu số nguyên
marketing_data["Income_changed"] = marketing_data["Income"].astype(int)

print("Số giá trị đã điền:", income_missing.sum())
print("Số giá trị thiếu còn lại:", marketing_data["Income"].isna().sum())

print("\nKiểu dữ liệu:")
print(marketing_data[["Income", "Income_changed"]].dtypes)

# Xem một số dòng vừa được điền
marketing_data.loc[
    income_missing, ["ID", "Income", "Income_changed"]
].head()
```

```python
# Kiểm tra giá trị thiếu trên bản trước khi điền
print("Giá trị thiếu trước khi loại bỏ:")
print(marketing_before_fill.isna().sum())

# Loại bỏ dòng có ít nhất một giá trị thiếu
marketing_data_withoutna = marketing_before_fill.dropna(how="any")

print("\nKích thước trước khi loại bỏ:", marketing_before_fill.shape)
print("Kích thước sau khi loại bỏ:", marketing_data_withoutna.shape)
print(
    "Số dòng đã loại bỏ:",
    len(marketing_before_fill) - len(marketing_data_withoutna)
)
print(
    "Tổng giá trị thiếu còn lại:",
    marketing_data_withoutna.isna().sum().sum()
)
```

### Nhận xét kết quả xử lý dữ liệu

- Dữ liệu ban đầu có 2.240 dòng và 29 cột. Sau khi chọn các cột
  theo lab, còn 11 cột.
- Không có dòng trùng hoàn toàn trong dữ liệu đã chọn.
  Vì vậy, drop_duplicates() không làm giảm số dòng.
- Thao tác xóa dòng index 1 và cột Year_Birth được thực hiện
  trên các bản riêng để minh họa, không thay đổi bảng đang dùng.
- Cột Teenhome_replaced có 1.158 khách hàng thuộc nhóm
  "has no teen" và 1.082 khách hàng thuộc nhóm "has teen".
  Cột Teenhome gốc được giữ lại.
- Có 24 giá trị thiếu ở Income. Theo bài mẫu, các giá trị này
  được điền bằng 0; đây là thao tác minh họa, không khẳng định
  thu nhập thực tế của khách hàng bằng 0.
- Cột Income_changed được chuyển sang kiểu số nguyên int64,
  trong khi Income vẫn có kiểu float64.
- Khi áp dụng dropna() trên bản trước khi điền, 24 dòng thiếu
  bị loại bỏ, còn 2.216 dòng và không còn giá trị thiếu.
- Điền giá trị thiếu và loại bỏ dòng thiếu là hai cách xử lý
  được minh họa trên các bản dữ liệu riêng.
